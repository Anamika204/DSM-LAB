```python
a = int(input("Enter a number: "))
b = int(input("Enter second number: "))
ch= int(input("Enter your choice: \n 1. Addition \n 2.substraction \n 3.Division \n 4.multiplication \n" ))
if ch == 1 :
    print (a + b)
elif ch == 2:
    print (a - b)
elif ch == 3:
    if b != 0:
        print (a/b)
    else:
        print ("can't divide by zero")
elif ch == 4:
        print (a*b)
    
else:
    print ("invalid choice")
```
