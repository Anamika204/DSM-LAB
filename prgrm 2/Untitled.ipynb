```python
import numpy as np
a = np.array([[2,4],[6,3]]);
b = np.array([[5,1],[8,4]]);
print ("matrix 1: \n ",a)
print ("matrix 2: \n ",b)
c = a + b
s = a - b
print ("sum:\n ",c)
print ("difference:\n ",s)
print ("multiplication:\n ", np.dot(a,b))
print ("transpose: \n", np.transpose(a))
```

```python
import numpy as np
a = np.array([[2,4],[6,3]]);
u,s,vt = np.linalg.svd(a)
n_components = 2
a_reconstructed = np.dot(u[:,:n_components],np.dot(np.diag(s[:n_components]), vt[:n_components,:]))
print ("Orginal matrix: ")
print(a)
print("\nReconsructed Maqtrix(with reduced dimensions):")
print(a_reconstructed)
```
